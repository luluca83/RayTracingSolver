```julia
function func(x::Vector{Float64})
    (x[1]-1)^2 + x[2]^2 + x[3]^4
end

function funcder(x::Vector{Float64})
     [ 2 * (x[1]-1) ; 2 * x[2] ; 4 * x[3]^3]
end

# How to measure the distance to zero of the gradient of f
function distance1(x::Vector{Float64})
    return x[1]^2 + x[2]^2 + x[3]^2
end

# How to evaluate the distance between two consecutive steps
function distance2(x::Vector{Float64})
    return x[1]^2 + x[2]^2 + x[3]^2
end
#problème

x0 = [0. ; 0. ; 0.]
step = 0.4
OptimGradient(func, funcder, x0, step, distance1, distance2)
```
